```python
import os
import pandas as pd

# csv dosyası dataframe olarak kaydedilir.
df = pd.read_csv(r"file_path")

# Dataframe içerisindeki en uzun ve en kısa peptit sekansları ile ortalama uzunluğu bulmak için fonksiyon:
def peptide_seq_stats(df):
    
    # NaN değerleri temizlenir ve DataFrame'in temiz kopyası oluşturulur. .copy() kullanmadığımda uyarı alıyorum.
    df_cleaned = df.dropna(subset=["sequence"]).copy()
    
    # Tüm sekanslar string türüne çevrilir.
    df_cleaned["sequence"] = df_cleaned["sequence"].astype(str)
    
    # Her bir sekansın uzunluğu hesaplanıp yeni bir sütun olarak ekler.
    df_cleaned["sequence_length"] = df_cleaned["sequence"].apply(len)
    
    # En uzun sekans bulunur.
    longest_sequence = df_cleaned.loc[df_cleaned["sequence_length"].idxmax()]
    
    # En kısa sekans bulunur.
    shortest_sequence = df_cleaned.loc[df_cleaned["sequence_length"].idxmin()]
    
    # Ortalama sekans uzunluğu hesaplanır.
    average_sequence_length = df_cleaned["sequence_length"].mean()
    
    return longest_sequence, shortest_sequence, average_sequence_length

# Fonksiyon çağrılır.
longest_sequence, shortest_sequence, average_sequence_length = peptide_seq_stats(df)

# Bulunan Series nesneleri Dataframe'e dönüştürülür ve transpoze edilir.
longest_sequence_df = longest_sequence.to_frame().T
shortest_sequence_df = shortest_sequence.to_frame().T

# Ortalama uzunluk için Dataframe oluşturulur. Tek değer olduğu için direk birleştirilemiyor.
average_sequence_length_df = pd.DataFrame({"Average_Length": [average_sequence_length]})

# Dataframe'ler birleştirilir.
sequences_df = pd.concat([longest_sequence_df, shortest_sequence_df, average_sequence_length_df], keys=["Longest", "Shortest", "Average"])

# Excel'e kaydedilir.
excel_path = "file_name.xlsx"
sequences_df.to_excel(excel_path, index=False)
```
