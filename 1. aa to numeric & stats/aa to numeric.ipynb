```python
import os
import pandas as pd

# csv dosyası Dataframe olarak yüklenir.
df = pd.read_csv(r"file_path")

# Her aa için bir sayı atanır, non-standard amino asitler de veri setinde olduğu için 26 tane... X ise "dummy" amino asit olarak kullanılıyor.
amino_acid_numeric_map = {"A": 1, "R": 2, "N": 3, "D": 4, "C": 5, "Q": 6, "E": 7, "G": 8, "H": 9, "I": 10, "L": 11, "K": 12, "M": 13, "F": 14, "P": 15, "O": 16, "S": 17, "U": 18, "T": 19, "W": 20, "Y": 21, "V": 22, "B": 23, "Z": 24, "X": 25, "J": 26}

# Amino asitleri sayısal değerlere dönüştürme fonksiyonu
def convert_sequence_to_numeric(sequence, mapping):
    # Eğer amino asit haritasında mevcut değilse, 25'e döndürerek sayısal dönüşümü gerçekleştirir.
    return [mapping.get(aa, 25) for aa in sequence]

# DataFrame'deki NaN değerleri temizler ve bir kopyasını oluşturur. "Boşluk" vs. değerlerden hata alıyorum.
df_cleaned = df.dropna(subset=['sequence']).copy()

# Sekansları büyük harfe çevirir. Bazı veri setlerinde küçük harf var.
df_cleaned['sequence'] = df_cleaned['sequence'].str.upper()

# Sekansları string olarak dönüştürür.
df_cleaned['sequence'] = df_cleaned['sequence'].astype(str)

# Sayısal değere dönüştürme fonksiyonu çağrılır.
df_cleaned['numeric_sequence'] = df_cleaned['sequence'].apply(lambda x: convert_sequence_to_numeric(x, amino_acid_numeric_map))

# csv dosyasına kaydetmek için dosya yolu
csv_path = "file_name.csv"

# Dataframe'i csv dosyası olarak kaydeder.
df_cleaned.to_csv(csv_path, index=False)
```
