```python
import pandas as pd
import numpy as np # tutorialda indirtiyor ama bunlar olmadan dene; scipy dahil
import matplotlib.pyplot as plt
from statistics import mean, stdev
from scipy import stats
from pepfun import *
```

```python
df = pd.read_csv("filtered_acp_sonuc.csv")
sequences = df["sequence"].tolist()
```

```python
charges = []
weights = []
logps = []
hydros = []

for seq in sequences:
    pep = peptide_sequence(seq)
    pep.compute_peptide_charges()
    pep.calculate_properties_from_mol()
    pep.calculate_properties_from_sequence()

    charges.append(pep.netCharge)
    weights.append(pep.mol_weight)
    logps.append(pep.mol_logp)
    hydros.append(pep.avg_hydro)
```

```python
results_df = pd.DataFrame({
    'sequence': sequences,
    'charge': charges,
    'weight': weights,
    'logP': logps,
    'hydrophobicity': hydros
})
results_df.to_csv('peptide_properties.csv', index=False)
```
