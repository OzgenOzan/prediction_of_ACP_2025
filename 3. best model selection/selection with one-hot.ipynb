```python
import os
import pandas as pd
import numpy as np
import ast
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, f1_score, roc_auc_score, recall_score
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

```python
# Veri seti yüklenir.
df = pd.read_csv(r"file_path")

# One_Hot_Encoded sütunundaki verileri NumPy dizisine dönüştürür ve düzleştirir.
one_hot_encoded_data = np.array(df['One_Hot_Encoded'].apply(ast.literal_eval).tolist())
# Düzleştirme işlemi (modellerin anlaması için düzleştirmek gerekiyor, yoksa hepsi aynı sonucu veriyor)
flattened_data = np.array([np.concatenate(peptide) for peptide in one_hot_encoded_data])

# Hedef değişken anticancer
y = df["anticancer"].values

# Veri setini, eğitim ve test setlerine ayırır.
X_train, X_test, y_train, y_test = train_test_split(flattened_data, y, test_size=0.2, random_state=42)
```

Random Forest

```python
# RF modelini oluşturur ve eğitir.
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)

# Test verisi üzerinde tahmin yap ve doğruluk oranını hesapla
rf_predictions = rf_model.predict(X_test)

# Accuracy
rf_accuracy = accuracy_score(y_test, rf_predictions)
print(f"Random Forest Doğruluk Oranı: {rf_accuracy}")

# Sensitivity ve Specificity
cm = confusion_matrix(y_test, rf_predictions)
sensitivity = cm[1, 1] / (cm[1, 1] + cm[1, 0])  # TP / (TP + FN)
specificity = cm[0, 0] / (cm[0, 0] + cm[0, 1])  # TN / (TN + FP)
print(f"Sensitivity: {sensitivity}")
print(f"Specificity: {specificity}")

# Precision
precision = precision_score(y_test, rf_predictions)
print(f"Precision: {precision}")

# F1 Score
f1 = f1_score(y_test, rf_predictions)
print(f"F1 Score: {f1}")

# ROC AUC Score
roc_auc = roc_auc_score(y_test, rf_model.predict_proba(X_test)[:, 1])
print(f"ROC AUC Score: {roc_auc}")
```

Logistic Regression

```python
# LR modelini oluşturur ve eğitir.
lr_model = LogisticRegression()
lr_model.fit(X_train, y_train)

# Test verisi üzerinde tahmin yap ve doğruluk oranını hesapla
lr_predictions = lr_model.predict(X_test)

# Accuracy
lr_accuracy = accuracy_score(y_test, lr_predictions)
print(f"Logistic Regression Doğruluk Oranı: {lr_accuracy}")

# Sensitivity ve Specificity
cm = confusion_matrix(y_test, lr_predictions)
sensitivity = cm[1, 1] / (cm[1, 1] + cm[1, 0])  # TP / (TP + FN)
specificity = cm[0, 0] / (cm[0, 0] + cm[0, 1])  # TN / (TN + FP)
print(f"Sensitivity: {sensitivity}")
print(f"Specificity: {specificity}")

# Precision
precision = precision_score(y_test, lr_predictions)
print(f"Precision: {precision}")

# F1 Score
f1 = f1_score(y_test, lr_predictions)
print(f"F1 Score: {f1}")

# ROC AUC Score
roc_auc = roc_auc_score(y_test, lr_model.predict_proba(X_test)[:, 1])
print(f"ROC AUC Score: {roc_auc}")
```

Support Vector Machine

```python
# SVM modelini oluşturur ve eğitir.
svm_model = SVC(probability=True) # ROC AUC hesaplamak için probability=True olması gerekiyormuş.
svm_model.fit(X_train, y_train)

# Test verisi üzerinde tahmin yap
svm_predictions = svm_model.predict(X_test)

# Accuracy
svm_accuracy = accuracy_score(y_test, svm_predictions)
print(f"SVM Doğruluk Oranı: {svm_accuracy}")

# Sensitivity ve Specificity
cm = confusion_matrix(y_test, svm_predictions)
sensitivity = cm[1, 1] / (cm[1, 1] + cm[1, 0])  # TP / (TP + FN)
specificity = cm[0, 0] / (cm[0, 0] + cm[0, 1])  # TN / (TN + FP)
print(f"Sensitivity: {sensitivity}")
print(f"Specificity: {specificity}")

# Precision
precision = precision_score(y_test, svm_predictions)
print(f"Precision: {precision}")

# F1 Score
f1 = f1_score(y_test, svm_predictions)
print(f"F1 Score: {f1}")

# ROC AUC Score
roc_auc = roc_auc_score(y_test, svm_model.predict_proba(X_test)[:, 1])
print(f"ROC AUC Score: {roc_auc}")
```

K-Nearest Neighbors (KNN)

```python
# KNN modelini oluşturur ve eğitir.
knn_model = KNeighborsClassifier()
knn_model.fit(X_train, y_train)

# Test verisi üzerinde tahmin yap
knn_predictions = knn_model.predict(X_test)

# Accuracy
knn_accuracy = accuracy_score(y_test, knn_predictions)
print(f"KNN Doğruluk Oranı: {knn_accuracy}")

# Sensitivity ve Specificity
cm = confusion_matrix(y_test, knn_predictions)
sensitivity = cm[1, 1] / (cm[1, 1] + cm[1, 0])  # TP / (TP + FN)
specificity = cm[0, 0] / (cm[0, 0] + cm[0, 1])  # TN / (TN + FP)
print(f"Sensitivity: {sensitivity}")
print(f"Specificity: {specificity}")

# Precision
precision = precision_score(y_test, knn_predictions)
print(f"Precision: {precision}")

# F1 Score
f1 = f1_score(y_test, knn_predictions)
print(f"F1 Score: {f1}")

# ROC AUC Score
# KNN modelinde ROC AUC hesaplamak için 'predict_proba' gerekiyor.
roc_auc = roc_auc_score(y_test, knn_model.predict_proba(X_test)[:, 1])
print(f"ROC AUC Score: {roc_auc}")
```
