```python
import pandas as pd

# CSV dosyasını okuma
df = pd.read_csv("acp_sonuc.csv")

# 'X' içermeyen satırları filtreleme
filtered_df = df[~df["sequence"].str.contains('X')]

# Filtrelenmiş DataFrame'i yeni bir CSV dosyası olarak kaydetme
filtered_df.to_csv("filtered_acp_sonuc.csv", index=False)
```
