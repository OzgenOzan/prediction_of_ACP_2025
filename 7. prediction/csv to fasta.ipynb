```python
import pandas as pd
import os

# Load your CSV file
df = pd.read_csv("file_name.csv")  # Replace with your CSV file path

# Open a new FASTA file to write the sequences
with open("file_name.fasta", "w") as fasta_file:
    for index, row in df.iterrows():
        # Write the header with anticancer property
        fasta_file.write(f'>seq{index}\n')
        # Write the sequence
        fasta_file.write(f'{row["sequence"]}\n')
```
